```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import matplotlib.pyplot as plt


# Keras has this dataset built-in, so we can load it directly into train and test splits
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()


X_train = X_train / 255.0
X_test = X_test / 255.0


model = Sequential([
  
    Flatten(input_shape=(28, 28)),          
    

    Dense(128, activation='relu'),          
    

    Dense(64, activation='relu'),           
    

    Dense(10, activation='softmax')         
])

model.compile(optimizer='adam', 
              loss='sparse_categorical_crossentropy', 
              metrics=['accuracy'])

print("--- Training the Neural Network ---")

history = model.fit(X_train, y_train, epochs=10, validation_split=0.2, batch_size=32)


test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nFinal Test Accuracy: {test_acc:.4f}")


plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss', color='teal', lw=2)
plt.plot(history.history['val_loss'], label='Validation Loss', color='orange', lw=2, linestyle='--')
plt.title('Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)


plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Training Accuracy', color='teal', lw=2)
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', color='orange', lw=2, linestyle='--')
plt.title('Model Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
