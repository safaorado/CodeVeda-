```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
col_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'Price']
```

```python
house_df = pd.read_csv("house Prediction Data Set.csv", sep=r'\s+', header=None, names=col_names)
house_df.head()
```

```python
print("Columns inthe dataset:", house_df.columns.tolist()
```

```python
house_df = house_df.dropna()

X_house = house_df.drop('Price', axis=1)
y_house = house_df['Price']

X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

lr_model = LinearRegression()
lr_model.fit(X_train_h, y_train_h)

predictions = lr_model.predict(X_test_h)
mse = mean_squared_error(y_test_h, predictions)
r2 = r2_score(y_test_h, predictions)

print("--- Model Evaluation ---")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared: {r2:.4f}")

print("\n--- Model Coefficients ---")
coefficients = pd.DataFrame({'Feature': X_house.columns, 'Coefficient': lr_model.coef_})
coefficients['Abs_Coefficient'] = coefficients['Coefficient'].abs()
coefficients = coefficients.sort_values(by='Abs_Coefficient', ascending=False).drop('Abs_Coefficient', axis=1)

display(coefficients.head())
```
